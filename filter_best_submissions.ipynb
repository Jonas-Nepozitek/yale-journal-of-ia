```python
df.rename(columns={'Assigned Submission Number': 'submission_number', 'Type of submission ': 'submission_type', 'Reader (please enter your preferred first name only)': 'reader', 'Is the book being reviewed published in the past 12 months?': 'is_recent_book_review'}, inplace=True)
```

```python
df = df.replace({'reader': {'Vincient ': 'Vincient', 'Michael ': 'Michael', 'Gabriel Falk': 'Gabriel'}})
```

```python
criteria = ["clarity", "substantiation", "methodology", "originality", "contribution", "writing_quality", "organization", "publishability"]

def zscore_group(df, cols, by):
    def z(g):
        return (g[cols] - g[cols].mean()) / g[cols].std(ddof=0)
    zed = df.groupby(by, group_keys=False).apply(z)
    zed.columns = [f"{c}_z" for c in cols]
    return pd.concat([df, zed], axis=1)

df = zscore_group(df, criteria, by="reader")
```

```python
article_scores = (
    df
    .groupby("submission_number")[[f"{c}_z" for c in criteria]]
    .mean()
    .reset_index()
)

# Optional: overall composite score
article_scores["overall_z"] = article_scores[[f"{c}_z" for c in criteria]].mean(axis=1)
```

```python
topn = (
    article_scores
    .sort_values("overall_z", ascending=False)
    .head(50)
)

topn = topn[['submission_number', 'overall_z']]

topn = topn.merge(df[["submission_type", "submission_number", "notes", "reader", "is_recent_book_review", "length"] + criteria].drop_duplicates(), on="submission_number", how="left")
```
